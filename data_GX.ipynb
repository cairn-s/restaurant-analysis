# Import Libraries

```python
from great_expectations.data_context import FileDataContext
import warnings
warnings.filterwarnings('ignore')
```

# Initialization

```python
context = FileDataContext.create(project_root_dir='./')
```

```python
datasource_name = 'transactions-csv'
datasource = context.sources.add_pandas(datasource_name)
asset_name = 'outlet_transactions'
path_to_data = 'P2M3_fauzan_abdurrochman_data_clean.csv'
```

```python
asset = datasource.add_csv_asset(asset_name, filepath_or_buffer=path_to_data)
batch_request = asset.build_batch_request()
```

```python
# Creat an expectation suite
expectation_suite_name = 'transactions-dataset'
context.add_or_update_expectation_suite(expectation_suite_name)

# Create a validator using above expectation suite
validator = context.get_validator(
    batch_request = batch_request,
    expectation_suite_name = expectation_suite_name
)

# Check the validator
validator.head()
```

# Data Expectation

```python
# - Expectaion : 1

validator.expect_column_values_to_be_unique('receiptnumber')
```

```python
# - Expectation : 2

validator.expect_column_values_to_be_in_set('outlet', ['SHOP001', 'SHOP002', 'SHOP003'])
```

```python
# - Expectation : 3

validator.expect_column_mean_to_be_between(column='useloyaltycard', min_value=0, max_value=1)
```

```python
# - Expectation : 4

validator.expect_column_values_to_be_in_type_list('tax', ['integer', 'float'])
```

```python
# - Expectation : 5

validator.expect_column_min_to_be_between(column='netsales', min_value=0, max_value=None)
```

```python
# - Expectation : 6

validator.expect_column_values_to_be_of_type('items', 'object')
```

```python
# - Expectation : 7

validator.expect_table_row_count_to_equal(value=53820)
```

All 7 Expectations returned `"success" : true`, we know our data meet our expectations
